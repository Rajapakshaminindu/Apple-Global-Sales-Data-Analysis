Apple Global Sales Data Analysis


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
my =  pd.read_csv("apple_global_sales_dataset.csv")
my.head()
```

```python
my.info()
my.isnull().sum()
```

```python
my.describe()
```

# **Rating the variables with Scatterplots**

```python
sns.relplot(x="unit_price_usd", y="discounted_price_usd", data=my)
```

**Conclustion 1 - There is a strong positive relationship between unit price and discounted price, indicating that discounted price increases proportionally with unit price. However, higher-priced products appear to receive larger absolute discounts, suggesting aggressive discount strategies for premium products.**

```python
sns.relplot(x="unit_price_usd", y="revenue_local_currency", hue="region" , data=my)
```

**Conclusion 2 - The Asian region generates relatively higher revenue at lower unit prices compared to other regions. This suggests stronger demand or higher sales volume in Asia for more affordable products.**

```python
sns.relplot(x="fx_rate_to_usd", y="revenue_local_currency", kind="line" , data=my)
```

**Conclusion 3 - Revenue in local currency tends to increase significantly when the exchange rate to USD reaches higher levels (around 16,000). This suggests that currency fluctuations strongly influence reported revenue in local currency. A higher exchange rate may inflate local currency revenue values.**
